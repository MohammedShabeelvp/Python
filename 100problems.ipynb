```python
# 2. Check if a string is a palindrome(case insenstive, ignore  non alphanumerics)
def is_pal(s):
    t_s = [ch.lower() for ch in s if ch.isalnum()]
    return t_s == t_s[::-1]

if is_pal("ak Ka"):
    print("It is a palindrome")

  
```

```python
# 3. Count the frequeny of item in a list.(Returns a dict)
lst = ["apple", "cherry","apple","banana","apple", "cherry"]

def lst_freq(l):
    d = dict([])
    for x in l:
        d[x] = d.get(x,0)+1
    return d
print(lst_freq(lst))
```

```python
# 5,Rotate a list by right k leg 
def rotate(a,k):
    n = len(a)
    if n == 0:
        return a
    k %= n
    a[:] = a[-k:]+a[:-k]
    return a
lst = [1,2,3,4,5]
rotate(lst, 2)
```

```python
# 7.Given String,Return the first non repeating characters(or None) 
def no_rep(s):
    from collections import Counter
    c = Counter(s)
    for ch in s:
        if c[ch] == 1: return ch
    return None

st ="rered"
no_rep(st)
```

```python
def sort_two_list(l1,l2):
    i = j = 0 
    out =[]
    while  i < len(l1) and j < len(l2):
        if l1[i] <= l2[j]:
            out.append(l1[i])
            i +=1
        else:
            out.append(l2[j])
            j +=1
    out.extend(l1[i:])
    out.extend(l2[j:])
    return out

l1 = [1,2,4]
l2 = [1,3,5]

sort_two_list(l1,l2)
```

```python
# 9. Given  a number n, compute n! iteratively (no recursion)
def fact(n):
    fact = 1
    for i in range(2,n+1):
        fact *= i
    return fact

fact(5)
```

```python
# 10. Remove duplicates from the list
def rem_duplicate(l):
    s, out = set(), []
    for x in l:
        if x not in s:
            s.add(x) ; out.append(x)
    return out

lst =["Apple", "Berry","Berry","Cherry"]
rem_duplicate(lst) 
```

```python
# 11.Find the maximum product of two numbers in an array.
def max_prod_arr(arr):
    arr.sort()
    if arr[0] * arr[1] >= arr[-1] * arr[-2]:
        return arr[0] * arr[1]
    else:
        return arr[-2] * arr[-1]
    
arr=[-10, -3, 5, 6]


max_prod_arr(arr)
```

```python
# 12. Longest substring without repeating characters
def long_sub(s):
    window =''
    for ch in s:
        if ch not in window:
            window= window + ch

    return window
s = "pwwkew"

long_sub(s)
```

```python
# 13. Maximum subarray sum  (kadane's algorithm)
def mss(nums):
    res = nums[0]

    for right in range(len(nums)):
        CurrSum = 0
        for j in range(right,len(nums)):
            CurrSum += nums[j]
            res = max(res, CurrSum)
    return res
nums = [5,4,-1,7,8]
mss(nums)
```

```python
def maxSubarraySum(arr):
    
    # Stores the result (maximum sum found so far)
    res = arr[0]
    
    # Maximum sum of subarray ending at current position
    maxEnding = arr[0]

    for i in range(1, len(arr)):
        
        # Either extend the previous subarray or start 
        # new from current element
        maxEnding = max(maxEnding + arr[i], arr[i])
        
        # Update result if the new subarray sum is larger
        res = max(res, maxEnding)
    
    return res

if __name__ == "__main__":
    arr = [-2,1,-3,4,-1,2,1,-5,4]
    print(maxSubarraySum(arr))
```

```python
# 14. Rotate Array Left by k
def rotate(nums,k):
    n = len(nums)
    if n == 0 or k == 0:
        return nums
    else:
        k = k % n
        new = nums[k:] + nums[:k]
        return new
    
if __name__ == "__main__":
    nums = [1,2,3,4,5]
    k = 2

    print(rotate(nums,k))
```

```python
# 15. Check if Two Strings are Rotations
def two_string_rotation_check(s1, s2):
    if len(s1) == len(s2):
        if s2 not in s1+s1:
            return False
        else:
            return True
    else:
        return False

s1 = "abcde"
s2 = "abcda"    

two_string_rotation_check(s1, s2)
```
